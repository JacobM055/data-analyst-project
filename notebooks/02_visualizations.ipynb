```python
# Import libraries needed for visualization
# pandas for data manipulation
# matplotlib for basic plotting
# seaborn for styled statistical charts
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the dataset
df = pd.read_csv('../data/Bank Customer Churn Prediction.csv')

# Set the visual style for all charts
# 'whitegrid' gives clean charts with grid lines
sns.set_style('whitegrid')

print("Libraries loaded and data ready!")
```

```python
# ============================================
# CHART 1: OVERALL CHURN RATE
# Shows the split between customers who stayed vs left
# ============================================

# Count how many customers stayed (0) and churned (1)
churn_counts = df['churn'].value_counts()

# Create labels and colors for the chart
labels = ['Stayed', 'Churned']
colors = ['#2ecc71', '#e74c3c']

# Create the figure - figsize sets the width and height in inches
fig, ax = plt.subplots(figsize=(8, 6))

# Create a bar chart
# churn_counts.values gets the actual numbers
# color sets the bar colors
ax.bar(labels, churn_counts.values, color=colors)

# Add title and axis labels
ax.set_title('Customer Churn Overview', fontsize=16, fontweight='bold')
ax.set_xlabel('Customer Status', fontsize=12)
ax.set_ylabel('Number of Customers', fontsize=12)

# Add the actual numbers on top of each bar
for i, v in enumerate(churn_counts.values):
    ax.text(i, v + 50, str(v), ha='center', fontsize=12)

# Clean up the chart before showing it
# Displays the chart
plt.tight_layout()
plt.show()
```

```python
# Chart 2: Churn Rate by Country

# Calculate churn rate by country
country_churn = df.groupby('country')['churn'].mean() * 100

# Create labels for each bar
labels = ['France', 'Germany', 'Spain']
colors = ["#0048FF", "#FF0000", "#FFD900"]

# Create the figure - figsize sets the width and height in inches
fig, ax = plt.subplots(figsize=(8, 6))

# Create a bar chart
# churn_counts.values gets the actual numbers
# color sets the bar colors
ax.bar(labels, country_churn.values, color=colors)

# Add title and axis labels
ax.set_title('Country Churn Rate', fontsize=16, fontweight='bold')
ax.set_xlabel('Country', fontsize=12)
ax.set_ylabel('Churn Percentage', fontsize=12)

# Add the actual numbers on top of each bar
for i, v in enumerate(country_churn.values):
   ax.text(i, v + 0.5, f"{v:.2f}%", ha='center', fontsize=12)

# Clean up the chart before showing it
# Displays the chart
plt.tight_layout()
plt.show()
```

```python
# Chart 3: Churn Rate by Number of Products

# Calculate churn rate by number of product
product_churn = df.groupby('products_number')['churn'].mean() * 100

# Create labels for each bar
labels = ['1', '2', '3', '4']
colors = ["#00FF77", "#E100FF", "#FFA600", "#FF0000"]

# Create the figure - figsize sets the width and height in inches
fig, ax = plt.subplots(figsize=(8, 6))

# Create a bar chart
# churn_counts.values gets the actual numbers
# color sets the bar colors
ax.bar(labels, product_churn.values, color=colors)

# Add title and axis labels
ax.set_title('Churn Rate by Product', fontsize=16, fontweight='bold')
ax.set_xlabel('# of Products', fontsize=12)
ax.set_ylabel('Churn Percentage', fontsize=12)

# Add the actual numbers on top of each bar
for i, v in enumerate(product_churn.values):
   ax.text(i, v + 1, f"{v:.2f}%", ha='center', fontsize=12)

# Clean up the chart before showing it
# Displays the chart
plt.tight_layout()
plt.show()
```

```python
# Chart 4: Active vs Inactive Churn Rate

# Calculate churn rate for active/inactive members
active_churn = df.groupby('active_member')['churn'].mean() * 100

# Create labels for each bar
labels = ['Inactive', 'Active']
colors = ["#FF0000", "#00FF0D"]

# Create the figure - figsize sets the width and height in inches
fig, ax = plt.subplots(figsize=(8, 6))

# Create a bar chart
# churn_counts.values gets the actual numbers
# color sets the bar colors
ax.bar(labels, active_churn.values, color=colors)

# Add title and axis labels
ax.set_title('Churn Rate for active/inactive members', fontsize=16, fontweight='bold')
ax.set_xlabel('Active vs Inactive', fontsize=12)
ax.set_ylabel('Churn Percentage', fontsize=12)

# Add the actual numbers on top of each bar
for i, v in enumerate(active_churn.values):
   ax.text(i, v + 1, f"{v:.2f}%", ha='center', fontsize=12)

# Clean up the chart before showing it
# Displays the chart
plt.tight_layout()
plt.show()
```

```python
# Chart 5: Age Distribution - Churned vs Stayed
# A histogram shows how age is spread across two groups

# Separate the data into two groups
stayed = df[df['churn'] == 0]['age']
churned = df[df['churn'] == 1]['age']

# Create the figure
fig, ax = plt.subplots(figsize=(10, 6))

# Plot two overlapping histograms
# alpha controls transparency so both are visible
# bins=30 divides age into 30 buckets
ax.hist(stayed, bins=30, alpha=0.6, color='#2ecc71', label='Stayed')
ax.hist(churned, bins=30, alpha=0.6, color='#e74c3c', label='Churned')

# Add title and labels
ax.set_title('Age Distribution: Churned vs Stayed', fontsize=16, fontweight='bold')
ax.set_xlabel('Age', fontsize=12)
ax.set_ylabel('Number of Customers', fontsize=12)

# Add a legend so we know which color is which
ax.legend(fontsize=12)

plt.tight_layout()
plt.show()
```
